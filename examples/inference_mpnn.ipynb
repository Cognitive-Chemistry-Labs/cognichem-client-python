```python
from dotenv import load_dotenv
import os

from cognichem_client import CogniChemClient


# Load API key from .env file in current working directory
load_dotenv()
API_KEY = os.getenv("API_KEY")

# Initialize the CogniChem client
client = CogniChemClient(API_KEY)
```

```python
# Run an inference job using the "Hydration free energy" MPNN model
result = client.inference.run(
    model_type="mpnn",
    model_name="Hydration free energy",
    payload={
        "input_data": ["CCO", "CCN", "CCC"],
        "input_format": "smiles",
        "return_contributions": True,
        "is_public_model": True
    }
)
result
```
