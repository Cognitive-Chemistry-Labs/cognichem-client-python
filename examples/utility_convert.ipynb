```python
from dotenv import load_dotenv
import os

from cognichem_client import CogniChemClient


# Load API key from .env file in current working directory
load_dotenv()
API_KEY = os.getenv("API_KEY")

# Initialize the CogniChem client
client = CogniChemClient(API_KEY)
```

```python
result = client.utility.run(
    utility_type="convert",
    payload={
        "input_data": "CCO",
        "input_format": "smiles",
        "output_format": "pdbblock",
        "generate_3d": True,
        "add_hydrogens": True
    }
)
result
```
